```python
actual_unreduced_close_pairs = []
for pair in reduced_close_pairs:
    actual_unreduced_close_pairs.append(functions.reduced_to_unreduced(IN_redux, pair, IN_all_seq_unreduced, start_index=1))
actual_reduced_close_pairs = []
for actual_un in actual_unreduced_close_pairs:
    actual_reduced_close_pairs.append(functions.unreduced_to_reduced(IN_redux, actual_un))
print("Actual unreduced close pairs:", actual_unreduced_close_pairs)
print("Actual reduced close pairs:", actual_reduced_close_pairs)

seen_unreduced_pairs = set()
filtered_unreduced = []
filtered_reduced = []

for unreduced_pair, reduced_pair in zip(actual_unreduced_close_pairs, actual_reduced_close_pairs):
    r_wt, _, r_mt = functions.split_pair(reduced_pair)

    if unreduced_pair in seen_unreduced_pairs or r_wt == r_mt:
        continue

    seen_unreduced_pairs.add(unreduced_pair)
    filtered_unreduced.append(unreduced_pair)
    filtered_reduced.append(reduced_pair)

actual_unreduced_close_pairs = filtered_unreduced
actual_reduced_close_pairs = filtered_reduced

print("Filtered unreduced close pairs:", actual_unreduced_close_pairs)
print("Filtered reduced close pairs:", actual_reduced_close_pairs)
    
```
