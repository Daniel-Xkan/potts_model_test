```python
import csv
import os

IN_DE_dict = DE_dict(IN_path + 'data/kn.her2.all')

with open(insti_file_path, 'r') as file:
    double_mutation_list = [line.strip() for line in file]
    # Prepare the output data
    output_data = []
    for pair in double_mutation_list:
        pair1, pair2 = pair.split('-')
        de1 = get_DE(IN_DE_dict, pair1, False, IN_redux_dict)
        de2 = get_DE(IN_DE_dict, pair2, False, IN_redux_dict)
        de1 = round(de1, 2) if de1 is not None else None
        de2 = round(de2, 2) if de2 is not None else None
        output_data.append([pair, de1, de2])
    # Write to TSV file
    in_output_dir = os.path.join(output_path, 'IN')
    os.makedirs(in_output_dir, exist_ok=True)
    output_file = os.path.join(in_output_dir, 'double_pairs_output.tsv')
    with open(output_file, 'w', newline='') as tsvfile:
        writer = csv.writer(tsvfile, delimiter='\t')
        writer.writerow(['Top Pairs', 'de1', 'de2'])  # Write header
        writer.writerows(output_data)
```
