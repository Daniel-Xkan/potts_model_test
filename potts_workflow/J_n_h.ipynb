```python
import numpy as np

import plotly.graph_objects as go

# Generate spiral points
num_points = 10
theta = np.linspace(0, 4 * np.pi, num_points)  # Angle for the spiral
z = np.linspace(0, 2, num_points)  # Height of the spiral
x = np.sin(theta)  # X-coordinates
y = np.cos(theta)  # Y-coordinates

# Define indices for red points
red_indices = [2, 8]

# Define colors for the points
colors = ['red' if i in red_indices else 'blue' for i in range(num_points)]

# Create the figure
fig = go.Figure()

# Add the spiral line
fig.add_trace(go.Scatter3d(
    x=x, y=y, z=z,
    mode='lines',
    line=dict(color='black', width=1),  # Thick black lines
    name='Spiral'
))

# Add the points
fig.add_trace(go.Scatter3d(
    x=x, y=y, z=z,
    mode='markers',
    marker=dict(size=8, color=colors),  # Color the points
    name='Points'
))

# Update layout
fig.update_layout(
    title='3D Spiral',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    )
)

# Add lines connecting each point to every other point
for i in range(num_points):
    for j in range(i + 1, num_points):
        color = 'red' if i in red_indices and j in red_indices else 'black'
        fig.add_trace(go.Scatter3d(
            x=[x[i], x[j]],
            y=[y[i], y[j]],
            z=[z[i], z[j]],
            mode='lines',
            line=dict(color=color, width=1.5),
            showlegend=False
        ))


fig.update_layout(
    scene=dict(
        xaxis=dict(visible=False),
        yaxis=dict(visible=False),
        zaxis=dict(visible=False)
    )
)
# Show the plot
fig.show()
```

```python
import numpy as np

import plotly.graph_objects as go

# Define the number of points and red indices
num_points = 10
red_indices = [2, 8]

# Generate x and y coordinates for the dots
x = np.linspace(0, 4, num_points)
y = np.linspace(0, 4, num_points)

# Define colors for the points
colors_x = ['red' if i in red_indices else 'blue' for i in range(num_points)]
colors_y = ['red' if i in red_indices else 'blue' for i in range(num_points)]

# Create the figure
fig = go.Figure()

# Add points on the x-axis
for i in range(num_points):
    fig.add_trace(go.Scatter3d(
        x=[x[i]], y=[0], z=[0],
        mode='markers',
        marker=dict(size=8, color=colors_x[i]),
        name=f'Point {i} on X-axis'
    ))

# Add points on the y-axis
for i in range(num_points):
    fig.add_trace(go.Scatter3d(
        x=[0], y=[y[i]], z=[0],
        mode='markers',
        marker=dict(size=8, color=colors_y[i]),
        name=f'Point {i} on Y-axis'
    ))

# Connect nearby dots on the x-axis
for i in range(num_points - 1):
    fig.add_trace(go.Scatter3d(
        x=[x[i], x[i + 1]],
        y=[0, 0],
        z=[0, 0],
        mode='lines',
        line=dict(color='black', width=2),
        showlegend=False
    ))

# Connect nearby dots on the y-axis
for i in range(num_points - 1):
    fig.add_trace(go.Scatter3d(
        x=[0, 0],
        y=[y[i], y[i + 1]],
        z=[0, 0],
        mode='lines',
        line=dict(color='black', width=2),
        showlegend=False
    ))

# Update layout
# fig.update_layout(
#     title='3D Plot with Dots on X and Y Axes',
#     scene=dict(
#         xaxis=dict(range=[0, 2], title='X', visible=False),
#         yaxis=dict(range=[0, 2], title='Y', visible=False),
#         zaxis=dict(range=[0, 2], title='Z', visible=False)
#     )
# )

# Generate random z values for the surface
z_surface = np.random.rand(num_points, num_points)
z_surface = np.random.uniform(-2, 0, (num_points, num_points))

# Add the surface
fig.add_trace(go.Surface(
    z=z_surface,
    x=np.tile(x, (num_points, 1)),
    y=np.tile(y, (num_points, 1)).T,
    colorscale='Viridis',
    name='Surface'
))
# Show the plot
fig.show()
```
