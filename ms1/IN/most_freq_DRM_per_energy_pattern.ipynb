```python
import os
import pandas as pd

# Define the input folders and the full sequence file
folders = ["data/compensate_out", "data/rescue_out", "data/flip_out"]
fullseq_file = "data/in.fullseq"

# Read the full sequence file to get the total number of sequences
with open(fullseq_file, 'r') as f:
    total_sequences = len(f.readlines())
print("Total sequences:", total_sequences)
# Initialize a dictionary to store frequencies for each folder
# Initialize a dictionary to store the most frequent DRM pair and its frequency for each folder
most_frequent_drm_pairs = {}

# Process each folder
for folder in folders:
    max_frequency = 0
    max_drm_pair = None
    
    for file in os.listdir(folder):
        if file.endswith(".tsv") and os.path.getsize(os.path.join(folder, file)) > 0:
            file_path = os.path.join(folder, file)
            drm_pair = os.path.splitext(file)[0]  # Extract DRM pair from file name
            
            # Read the TSV file, skip if the file is empty
            try:
                df = pd.read_csv(file_path, sep='\t')
            except pd.errors.EmptyDataError:
                continue
            
            # Calculate the frequency for this file
            frequency = len(df) / total_sequences
            
            # Update the most frequent DRM pair for the folder
            if frequency > max_frequency:
                max_frequency = frequency
                max_drm_pair = drm_pair
    
    # Store the most frequent DRM pair and its frequency for the folder
    most_frequent_drm_pairs[folder] = (max_drm_pair, max_frequency)

# Print the most frequent DRM pair and its frequency for each folder
for folder, (drm_pair, frequency) in most_frequent_drm_pairs.items():
    print(f"Folder: {folder}, Most Frequent DRM Pair: {drm_pair}, Frequency: {frequency}")
```
