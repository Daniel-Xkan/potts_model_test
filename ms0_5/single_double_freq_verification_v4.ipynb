```python

IN_de_dict = {}
IN_de_reduced_to_unreduced = {}
IN_dde = {}
IN_dde_unreduced = {}
for i in tqdm(range(0, IN_max_position-IN_min_position+1), desc="IN Progress"):
    pos1 = i + 1
    wt1 = IN_consensus_seq[i]
    for aa in ['A', 'B', 'C', 'D']:
        if aa == wt1:
            continue
        mutation_pair1 = wt1 + str(pos1) + aa
        mutation_pair1_unreduced = functions.reduced_to_unreduced(
            IN_redux, mutation_pair1, IN_all_seq_unreduced, IN_min_position
        )
        de = functions.calculate_delta_e(mutation_pair1, IN_consensus_seq, IN_J, IN_min_position, IN_max_position)
        IN_de_dict[mutation_pair1] = float(de)
        IN_de_reduced_to_unreduced[mutation_pair1] = mutation_pair1_unreduced
# IN:
for i in tqdm(range(0, IN_max_position-IN_min_position+1), desc="IN Progress"):
    pos1 = i+1
    wt1 = IN_consensus_seq[i]
    for aa in ['A','B','C','D']:
        if aa == wt1:
            continue
        mutation_pair1 = wt1 + str(pos1) + aa
        for j in range(i+1, IN_max_position-IN_min_position+1):
            pos2 = j+1
            wt2 = IN_consensus_seq[j]
            for aa in ['A','B','C','D']:
                if aa == wt2:
                    continue
                mutation_pair2 = wt2 + str(pos2) + aa
                de12 = functions.calculate_delta_e_double(mutation_pair1, mutation_pair2, IN_consensus_seq, IN_J, IN_min_position, IN_max_position)
                dde = de12 - IN_de_dict[mutation_pair1] - IN_de_dict[mutation_pair2]
                IN_dde[(mutation_pair1, mutation_pair2)] = float(dde)

                pair1_unreduced = IN_de_reduced_to_unreduced.get(mutation_pair1, mutation_pair1)
                pair2_unreduced = IN_de_reduced_to_unreduced.get(mutation_pair2, mutation_pair2)
                IN_dde_unreduced[(pair1_unreduced, pair2_unreduced)] = float(dde)

pd.DataFrame(
    [(k, v) for k, v in IN_de_dict.items()],
    columns=["mutation", "delta_e"]
).to_csv("IN_de_dict.csv", index=False)

pd.DataFrame(
    [(k, v) for k, v in IN_de_reduced_to_unreduced.items()],
    columns=["mutation", "unreduced_mutation"]
).to_csv("IN_de_reduced_to_unreduced.csv", index=False)

pd.DataFrame(
    [(str(k), v) for k, v in IN_dde.items()],
    columns=["mutation_pair", "dde"]
).to_csv("IN_dde.csv", index=False)

pd.DataFrame(
    [(str(k), v) for k, v in IN_dde_unreduced.items()],
    columns=["mutation_pair", "dde"]
).to_csv("IN_dde_unreduced.csv", index=False)

print("CSV files saved successfully.")
```

```python
#instead do the follwing: get 100 double pairs where each single pair has >1 observed frequency. first column is the double mutation pair, second column: pair1_DRP, third: pos2DRP 1 if in drug resisitant positions, fourth pair1_DRM (leave it empty for now), fifth: pair2_DRM (leave it empty for now).
import ast
import re
length_threshold = 50

csv_candidates = ["IN_dde_dict.csv", "IN_dde.csv"]
for csv_file in csv_candidates:
    try:
        in_dde_df = pd.read_csv(csv_file)
        print(f"Loaded: {csv_file}")
        break
    except FileNotFoundError:
        in_dde_df = None

if in_dde_df is None:
    raise FileNotFoundError("Neither IN_dde_dict.csv nor IN_dde.csv was found.")

def parse_mutation(mut):
    m = re.match(r"^([A-Z])(\d+)([A-Z])$", mut)
    if not m:
        raise ValueError(f"Cannot parse mutation string: {mut}")
    wt, pos, mt = m.groups()
    return wt, int(pos), mt

def single_mutant_frequency(seqs, mutation, min_pos=1):
    _, pos, mt = parse_mutation(mutation)
    idx = pos - min_pos
    return sum(1 for s in seqs if len(s) > idx and s[idx] == mt)

records = []

for _, row in in_dde_df.sort_values("dde", ascending=False).iterrows():
    m1_red, m2_red = ast.literal_eval(row["mutation_pair"])

    m1 = IN_de_reduced_to_unreduced.get(m1_red, m1_red)
    m2 = IN_de_reduced_to_unreduced.get(m2_red, m2_red)

    f1 = single_mutant_frequency(IN_all_seq_unreduced, m1)
    f2 = single_mutant_frequency(IN_all_seq_unreduced, m2)

    if f1 > 1 and f2 > 1:
        _, pos1, _ = parse_mutation(m1)
        _, pos2, _ = parse_mutation(m2)

        records.append({
            "double_mutation_pair": f"{m1}-{m2}",
            "pos1_DRP": 1 if pos1 in INSTI_DRM_positions else 0,
            "pos2_DRP": 1 if pos2 in INSTI_DRM_positions else 0,
            "both_DRP": 1 if (pos1 in INSTI_DRM_positions and pos2 in INSTI_DRM_positions) else 0,
            "pair1_DRM": 1 if m1 in INSTI_DRM_mutations else 0,
            "pair2_DRM": 1 if m2 in INSTI_DRM_mutations else 0,
            "both_DRM": 1 if (m1 in INSTI_DRM_mutations and m2 in INSTI_DRM_mutations) else 0,
        })

    if len(records) == length_threshold:
        break


summary_df = pd.DataFrame(records)
summary_df.to_csv(f"IN_top_{length_threshold}_double_pairs_filtered.csv", index=False)
print(f"Saved: IN_top_{length_threshold}_double_pairs_filtered.csv")

# Calculate and print statistics
total_pairs = len(summary_df)

both_drp_mask = summary_df["both_DRP"] == 1
either_drp_mask = (summary_df["pos1_DRP"] == 1) | (summary_df["pos2_DRP"] == 1)
both_drm_mask = summary_df["both_DRM"] == 1
either_drm_mask = (summary_df["pair1_DRM"] == 1) | (summary_df["pair2_DRM"] == 1)

both_drp_count = both_drp_mask.sum()
either_drp_count = either_drp_mask.sum()
both_drm_count = both_drm_mask.sum()
either_drm_count = either_drm_mask.sum()

both_drp_pct = (both_drp_count / total_pairs) * 100 if total_pairs else 0
either_drp_pct = (either_drp_count / total_pairs) * 100 if total_pairs else 0
both_drm_pct = (both_drm_count / total_pairs) * 100 if total_pairs else 0
either_drm_pct = (either_drm_count / total_pairs) * 100 if total_pairs else 0

print(f"Total pairs: {total_pairs}")
print(f"Pairs with both positions as DRP: {both_drp_count} ({both_drp_pct:.1f}%)")
print(f"Pairs with either position as DRP: {either_drp_count} ({either_drp_pct:.1f}%)")
print(f"Pairs with both mutations as DRM: {both_drm_count} ({both_drm_pct:.1f}%)")
print(f"Pairs with either mutation as DRM: {either_drm_count} ({either_drm_pct:.1f}%)")

print("\nPairs with both positions as DRP:")
print(summary_df.loc[both_drp_mask, "double_mutation_pair"].tolist())

print("\nPairs with either position as DRP:")
print(summary_df.loc[either_drp_mask, "double_mutation_pair"].tolist())

print("\nPairs with both mutations as DRM:")
print(summary_df.loc[both_drm_mask, "double_mutation_pair"].tolist())

print("\nPairs with either mutation as DRM:")
print(summary_df.loc[either_drm_mask, "double_mutation_pair"].tolist())
```

```python
import ast
import re
length_threshold = 50

csv_candidates = ["PR_dde_dict.csv", "PR_dde.csv"]
for csv_file in csv_candidates:
    try:
        pr_dde_df = pd.read_csv(csv_file)
        print(f"Loaded: {csv_file}")
        break
    except FileNotFoundError:
        pr_dde_df = None

if pr_dde_df is None:
    raise FileNotFoundError("Neither PR_dde_dict.csv nor PR_dde.csv was found.")

def parse_mutation(mut):
    m = re.match(r"^([A-Z])(\d+)([A-Z])$", mut)
    if not m:
        raise ValueError(f"Cannot parse mutation string: {mut}")
    wt, pos, mt = m.groups()
    return wt, int(pos), mt

def single_mutant_frequency(seqs, mutation, min_pos=1):
    _, pos, mt = parse_mutation(mutation)
    idx = pos - min_pos
    return sum(1 for s in seqs if len(s) > idx and s[idx] == mt)

records = []

for _, row in pr_dde_df.sort_values("dde", ascending=False).iterrows():
    m1_red, m2_red = ast.literal_eval(row["mutation_pair"])

    m1 = PR_de_reduced_to_unreduced.get(m1_red, m1_red)
    m2 = PR_de_reduced_to_unreduced.get(m2_red, m2_red)

    f1 = single_mutant_frequency(PR_all_seq_unreduced, m1, PR_min_position)
    f2 = single_mutant_frequency(PR_all_seq_unreduced, m2, PR_min_position)

    if f1 > 1 and f2 > 1:
        _, pos1, _ = parse_mutation(m1)
        _, pos2, _ = parse_mutation(m2)

        records.append({
            "double_mutation_pair": f"{m1}-{m2}",
            "pos1_DRP": 1 if pos1 in PI_DRM_positions else 0,
            "pos2_DRP": 1 if pos2 in PI_DRM_positions else 0,
            "both_DRP": 1 if (pos1 in PI_DRM_positions and pos2 in PI_DRM_positions) else 0,
            "pair1_DRM": 1 if m1 in PI_DRM_mutations else 0,
            "pair2_DRM": 1 if m2 in PI_DRM_mutations else 0,
            "both_DRM": 1 if (m1 in PI_DRM_mutations and m2 in PI_DRM_mutations) else 0,
        })

    if len(records) == length_threshold:
        break

summary_df = pd.DataFrame(records)
summary_df.to_csv(f"PR_top_{length_threshold}_double_pairs_filtered.csv", index=False)
print(f"Saved: PR_top_{length_threshold}_double_pairs_filtered.csv")

total_pairs = len(summary_df)

both_drp_mask = summary_df["both_DRP"] == 1
either_drp_mask = (summary_df["pos1_DRP"] == 1) | (summary_df["pos2_DRP"] == 1)
both_drm_mask = summary_df["both_DRM"] == 1
either_drm_mask = (summary_df["pair1_DRM"] == 1) | (summary_df["pair2_DRM"] == 1)

both_drp_count = both_drp_mask.sum()
either_drp_count = either_drp_mask.sum()
both_drm_count = both_drm_mask.sum()
either_drm_count = either_drm_mask.sum()

both_drp_pct = (both_drp_count / total_pairs) * 100 if total_pairs else 0
either_drp_pct = (either_drp_count / total_pairs) * 100 if total_pairs else 0
both_drm_pct = (both_drm_count / total_pairs) * 100 if total_pairs else 0
either_drm_pct = (either_drm_count / total_pairs) * 100 if total_pairs else 0

print(f"Total pairs: {total_pairs}")
print(f"Pairs with both positions as DRP: {both_drp_count} ({both_drp_pct:.1f}%)")
print(f"Pairs with either position as DRP: {either_drp_count} ({either_drp_pct:.1f}%)")
print(f"Pairs with both mutations as DRM: {both_drm_count} ({both_drm_pct:.1f}%)")
print(f"Pairs with either mutation as DRM: {either_drm_count} ({either_drm_pct:.1f}%)")

print("\nPairs with both positions as DRP:")
print(summary_df.loc[both_drp_mask, "double_mutation_pair"].tolist())

print("\nPairs with either position as DRP:")
print(summary_df.loc[either_drp_mask, "double_mutation_pair"].tolist())

print("\nPairs with both mutations as DRM:")
print(summary_df.loc[both_drm_mask, "double_mutation_pair"].tolist())

print("\nPairs with either mutation as DRM:")
print(summary_df.loc[either_drm_mask, "double_mutation_pair"].tolist())
```
